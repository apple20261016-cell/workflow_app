```python
import streamlit as st

# ページの設定
st.set_page_config(page_title="社内ワークフローシステム", page_icon="🔐", layout="centered")

# セッション状態（ログイン状態）の初期化
if "authenticated" not in st.session_state:
    st.session_state.authenticated = False

def check_password():
    """パスワード認証の判定"""
    # st.secrets からパスワードを取得（未設定時はテスト用の 'secret123'）
    correct_password = st.secrets.get("PASSWORD", "secret123")
    
    password_input = st.text_input("パスワードを入力してください", type="password")
    if st.button("ログイン"):
        if password_input == correct_password:
            st.session_state.authenticated = True
            st.rerun()
        else:
            st.error("パスワードが違います。")

# 1. 未認証の場合はログイン画面を表示
if not st.session_state.authenticated:
    st.title("🔐 社内ワークフローシステム")
    st.info("このアプリを利用するにはパスワードが必要です。")
    check_password()

# 2. 認証済みの場合はメインのワークフロー画面を表示
else:
    st.title("📋 簡易ワークフロー申請フォーム")
    st.write("必要事項を入力して申請してください。")

    with st.form("workflow_form"):
        applicant = st.text_input("申請者氏名")
        department = st.selectbox("所属部署", ["営業部", "開発部", "総務部", "マーケティング部"])
        title = st.text_input("申請件名")
        category = st.selectbox("申請種別", ["経費精算", "備品購入", "出張申請", "その他"])
        amount = st.number_input("金額（円）", min_value=0, step=1000)
        reason = st.text_area("申請理由・詳細")
        
        submitted = st.form_submit_button("申請を送信する")
        if submitted:
            if applicant and title:
                # ここにGoogleスプレッドシートやデータベースへ保存する処理を記述
                st.success(f"【送信完了】{applicant} さんからの申請（{title}）を受け付けました！")
            else:
                st.warning("申請者氏名と件名は必須です。")

    # サイドバーにログアウトボタン等を配置
    with st.sidebar:
        st.write(f"ログイン中ユーザー")
        if st.button("ログアウト"):
            st.session_state.authenticated = False
            st.rerun()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. パスワード認証用のウィジェット定義
password_input = widgets.Password(
    description='パスワード:', 
    placeholder='secret123 と入力',
    style={'description_width': 'initial'}
)
login_button = widgets.Button(
    description='ログイン', 
    button_style='primary',
    icon='key'
)
login_output = widgets.Output()

def handle_login(b):
    with login_output:
        clear_output()
        # パスワードの判定（初期値は secret123）
        if password_input.value == "secret123":
            print("✅ ログイン成功！申請フォームを表示します...")
            display_workflow_form()
        else:
            print("❌ パスワードが違います。('secret123' と入力してください)")

login_button.on_click(handle_login)

# 最初にログイン画面を表示
print("🔐 社内ワークフローシステム (認証画面)")
display(widgets.VBox([password_input, login_button, login_output]))


# 2. 認証成功後に表示されるメインフォームの関数
def display_workflow_form():
    # 画面をクリアしてフォームを描画
    clear_output()
    
    display(widgets.HTML("<h3>📋 簡易ワークフロー申請フォーム</h3>"))
    
    applicant = widgets.Text(description='申請者氏名:', placeholder='例: 山田 太郎', style={'description_width': 'initial'})
    department = widgets.Dropdown(options=['営業部', '開発部', '総務部', 'マーケティング部'], description='所属部署:', style={'description_width': 'initial'})
    title = widgets.Text(description='申請件名:', placeholder='例: 27インチモニター購入', style={'description_width': 'initial'})
    category = widgets.Dropdown(options=['経費精算', '備品購入', '出張申請', 'その他'], description='申請種別:', style={'description_width': 'initial'})
    amount = widgets.IntText(value=0, description='金額(円):', style={'description_width': 'initial'})
    reason = widgets.Textarea(description='申請理由:', placeholder='詳細を入力してください', style={'description_width': 'initial'})
    
    submit_button = widgets.Button(description='申請を送信する', button_style='success', icon='check')
    output = widgets.Output()
    
    def handle_submit(b):
        with output:
            clear_output()
            if applicant.value and title.value:
                print("----------------------------------------")
                print(f"🎉 【送信完了】以下の内容で受け付けました！")
                print(f"・申請者: {applicant.value} ({department.value})")
                print(f"・件名: {title.value} [{category.value}]")
                print(f"・金額: {amount.value:,} 円")
                print(f"・理由: {reason.value}")
                print("----------------------------------------")
            else:
                print("⚠️ 「申請者氏名」と「申請件名」は必須です。")
                
    submit_button.on_click(handle_submit)
    
    # ウィジェットを縦に並べて表示
    form_box = widgets.VBox([
        applicant, department, title, category, amount, reason, submit_button, output
    ])
    display(form_box)
```
